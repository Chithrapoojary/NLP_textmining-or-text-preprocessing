```python
import nltk
```

```python
nltk.download("popular")
```

```python
paragraph="Speaking to reporters after the successful launch, ISRO Chairman S. Somanath said the next 42 days are crucial. “As per the nominal programme, we will have five earth-bound manoeuvres [that] will end on July 31. After that we have the trans-lunar insertion, [which] will take place on August 1. After that, it will be captured [by the] moon.This will be followed by the separation of the propulsion module and the lander module on August 17. “The landing is currently planned on August 23 at 5.47 pm IST, if everything goes as per plan,” he added"
```

```python
print(paragraph)
```

```python
#sentence level tokenization
sentences=nltk.sent_tokenize(paragraph)
```

```python
print(sentences)
```

```python
#Word level tokenization
words=nltk.word_tokenize(paragraph)
```

```python
print(words)
```

```python
#Stemming
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
from nltk.corpus import stopwords
stopwords.words("english")
```

```python
for i in range(len(sentences)):
    words_stem=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences_stem=' '.join(words_stem)
```

```python
print(sentences_stem)
```

```python
#lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words_new=[lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    sentences_new=' '.join(words_new)                                                                                         
```

```python
print(sentences_new)
```
